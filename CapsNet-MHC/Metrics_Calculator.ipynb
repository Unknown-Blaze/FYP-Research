## Import require libraries

```python
from sklearn.metrics import roc_auc_score, auc
from sklearn.metrics import precision_recall_curve
from sklearn.metrics import confusion_matrix
import numpy as np
from collections import Counter
```

## Read Results

```python
from os import readlink
f = open('./codes/Anthem_codes/Anthem_test/weekly_result.txt',"r")
lines = f.readlines()
y_prob8, y_prob9, y_prob10, y_prob11, y_prob12, y_prob13, y_prob14 = [], [], [], [], [], [], []
y_true8, y_true9, y_true10, y_true11, y_true12, y_true13, y_true14 = [], [], [], [], [], [], []
lines.pop(0)
plot = []
for x in lines:
#     if(x.split('\t')[0]) == 'HLA-A*01:01':
        if (x.split('\t')[3]) == '8':
            y_prob8.append(x.split('\t')[4])
            y_true8.append(x.split('\t')[2])
        if (x.split('\t')[3]) == '9':
            y_prob9.append(x.split('\t')[4])
            y_true9.append(x.split('\t')[2])
        if (x.split('\t')[3]) == '10':
            y_prob10.append(x.split('\t')[4])
            y_true10.append(x.split('\t')[2])
        if (x.split('\t')[3]) == '11':
            y_prob11.append(x.split('\t')[4])
            y_true11.append(x.split('\t')[2])
        if (x.split('\t')[3]) == '12':
            y_prob12.append(x.split('\t')[4])
            y_true12.append(x.split('\t')[2])
        if (x.split('\t')[3]) == '13':
            y_prob13.append(x.split('\t')[4])
            y_true13.append(x.split('\t')[2])
        if (x.split('\t')[3]) == '14':
            y_prob14.append(x.split('\t')[4])
            y_true14.append(x.split('\t')[2])
        
f.close()
y_prob8 = [float(num) for num in y_prob8]
y_true8 = [float(num) for num in y_true8]
y_pred8 = [round(num) for num in y_prob8]

y_prob9 = [float(num) for num in y_prob9]
y_true9 = [float(num) for num in y_true9]
y_pred9 = [round(num) for num in y_prob9]

y_prob10 = [float(num) for num in y_prob10]
y_true10 = [float(num) for num in y_true10]
y_pred10 = [round(num) for num in y_prob10]

y_prob11 = [float(num) for num in y_prob11]
y_true11 = [float(num) for num in y_true11]
y_pred11 = [round(num) for num in y_prob11]

y_prob12 = [float(num) for num in y_prob12]
y_true12 = [float(num) for num in y_true12]
y_pred12 = [round(num) for num in y_prob12]

y_prob13 = [float(num) for num in y_prob13]
y_true13 = [float(num) for num in y_true13]
y_pred13 = [round(num) for num in y_prob13]

y_prob14 = [float(num) for num in y_prob14]
y_true14 = [float(num) for num in y_true14]
y_pred14 = [round(num) for num in y_prob14]

plot.append(len(y_true8))
plot.append(len(y_true9))
plot.append(len(y_true10))
plot.append(len(y_true11))
plot.append(len(y_true12))
plot.append(len(y_true13))
plot.append(len(y_true14))
```

## 8-mer

```python
tn, fp, fn, tp = confusion_matrix(y_true8, y_pred8, labels = [0, 1]).ravel().tolist()
accuracy = (tp+tn)/(tn+fp+fn+tp)
try:
    mcc = ((tp*tn) - (fn*fp)) / np.sqrt(np.float((tp+fn)*(tn+fp)*(tp+fp)*(tn+fn)))
except:
    print('MCC Error: ', (tp+fn)*(tn+fp)*(tp+fp)*(tn+fn))
    mcc = np.nan
sensitivity = tp/(tp+fn)
specificity = tn/(tn+fp)

try:
    recall = tp / (tp+fn)
except:
    recall = np.nan

try:
    precision = tp / (tp+fp)
except:
    precision = np.nan

try: 
    f1 = 2*precision*recall / (precision+recall)
except:
    f1 = np.nan

roc_auc = roc_auc_score(y_true8, y_prob8)
prec, reca, _ = precision_recall_curve(y_true8, y_prob8)
aupr = auc(reca, prec)

print('tn = {}, fp = {}, fn = {}, tp = {}'.format(tn, fp, fn, tp))
print('y_pred8: 0 = {} | 1 = {}'.format(Counter(y_pred8)[0], Counter(y_pred8)[1]))
print('y_true8: 0 = {} | 1 = {}'.format(Counter(y_true8)[0], Counter(y_true8)[1]))
print('auc={:.4f}|sn={:.4f}|sp={:.4f}|acc={:.4f}|mcc={:.4f}'.format(roc_auc, sensitivity, specificity, accuracy, mcc))
print('precision={:.4f}|recall={:.4f}|f1={:.4f}|aupr={:.4f}'.format(precision, recall, f1, aupr))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import roc_curve, precision_recall_curve, auc
import numpy as np

print("Data aggregation cell executed. Variables are ready for plotting.")

# --- 1. Aggregate all the calculated metrics into a dictionary ---
# This data is taken directly from the output of your cells.
metrics_by_length = {
    '8-mer':  {'auc': 0.9844, 'f1': 0.9416, 'mcc': 0.8834},
    '9-mer':  {'auc': 0.9864, 'f1': 0.9486, 'mcc': 0.8969},
    '10-mer': {'auc': 0.9800, 'f1': 0.9346, 'mcc': 0.8690},
    '11-mer': {'auc': 0.9786, 'f1': 0.9294, 'mcc': 0.8606},
    '12-mer': {'auc': 0.9469, 'f1': 0.8700, 'mcc': 0.7463},
    '13-mer': {'auc': 0.9234, 'f1': 0.8348, 'mcc': 0.6742},
    '14-mer': {'auc': 0.8713, 'f1': 0.7931, 'mcc': 0.5772}
}

# --- 2. Aggregate the raw data lists for curve and matrix plotting ---
y_true_all = {
    '8-mer': y_true8, '9-mer': y_true9, '10-mer': y_true10, '11-mer': y_true11,
    '12-mer': y_true12, '13-mer': y_true13, '14-mer': y_true14
}
y_pred_all = {
    '8-mer': y_pred8, '9-mer': y_pred9, '10-mer': y_pred10, '11-mer': y_pred11,
    '12-mer': y_pred12, '13-mer': y_pred13, '14-mer': y_pred14
}
y_prob_all = {
    '8-mer': y_prob8, '9-mer': y_prob9, '10-mer': y_prob10, '11-mer': y_prob11,
    '12-mer': y_prob12, '13-mer': y_prob13, '14-mer': y_prob14
}
```

```python
# Extract data for plotting from the aggregated dictionary
labels = list(metrics_by_length.keys())
auc_scores = [metrics['auc'] for metrics in metrics_by_length.values()]
f1_scores = [metrics['f1'] for metrics in metrics_by_length.values()]
mcc_scores = [metrics['mcc'] for metrics in metrics_by_length.values()]

x = np.arange(len(labels))  # the label locations
width = 0.25  # the width of the bars

# Create the plot
fig, ax = plt.subplots(figsize=(16, 8))
rects1 = ax.bar(x - width, auc_scores, width, label='AUC', color='cornflowerblue')
rects2 = ax.bar(x, f1_scores, width, label='F1-Score', color='salmon')
rects3 = ax.bar(x + width, mcc_scores, width, label='MCC', color='mediumseagreen')

# Add some text for labels, title and axes ticks
ax.set_ylabel('Scores', fontsize=14)
ax.set_title('Model Performance Comparison by Peptide Length', fontsize=16)
ax.set_xticks(x)
ax.set_xticklabels(labels, fontsize=12)
ax.legend(fontsize=12)
ax.grid(axis='y', linestyle='--', alpha=0.7)

# Add value labels on top of each bar
ax.bar_label(rects1, padding=3, fmt='%.3f')
ax.bar_label(rects2, padding=3, fmt='%.3f')
ax.bar_label(rects3, padding=3, fmt='%.3f')

# Set y-axis limits to provide some space at the top
ax.set_ylim(0, 1.15)

fig.tight_layout()
plt.show()
```

```python
# Create a figure with two subplots (1 row, 2 columns)
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(18, 8))

# --- Subplot 1: ROC Curves ---
ax1.plot([0, 1], [0, 1], 'k--', label='Random Guess') # Dashed diagonal
for key, y_true_list in y_true_all.items():
    y_prob_list = y_prob_all[key]
    fpr, tpr, _ = roc_curve(y_true_list, y_prob_list)
    roc_auc = metrics_by_length[key]['auc'] # Get pre-calculated AUC
    ax1.plot(fpr, tpr, label=f'{key} (AUC = {roc_auc:.3f})', lw=2)

ax1.set_xlabel('False Positive Rate', fontsize=12)
ax1.set_ylabel('True Positive Rate (Recall)', fontsize=12)
ax1.set_title('Receiver Operating Characteristic (ROC) Curves', fontsize=14)
ax1.legend(loc='lower right', fontsize=10)
ax1.grid(True)

# --- Subplot 2: Precision-Recall Curves ---
for key, y_true_list in y_true_all.items():
    y_prob_list = y_prob_all[key]
    precision, recall, _ = precision_recall_curve(y_true_list, y_prob_list)
    pr_auc = auc(recall, precision)
    ax2.plot(recall, precision, label=f'{key} (AUPR = {pr_auc:.3f})', lw=2)

ax2.set_xlabel('Recall', fontsize=12)
ax2.set_ylabel('Precision', fontsize=12)
ax2.set_title('Precision-Recall Curves', fontsize=14)
ax2.legend(loc='lower left', fontsize=10)
ax2.grid(True)

plt.suptitle('Model Diagnostic Curves by Peptide Length', fontsize=18)
plt.tight_layout(rect=[0, 0, 1, 0.96]) # Adjust layout to make room for suptitle
plt.show()
```
